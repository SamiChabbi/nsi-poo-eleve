# 1. Pourquoi des objets ?

Un ninja a un **nom**, un **village** et une réserve de **chakra** (son énergie). Sans objets, on pourrait le ranger dans une liste :

````python
ninja = ["Kenji", "Vent", 50]
print(ninja[0])   # le nom
print(ninja[2])   # le chakra
````

Ça marche, mais `ninja[2]` ne dit pas ce que c'est : il faut se souvenir que la case 2 est le chakra. Et rien n'empêche d'écrire n'importe quoi dedans.

Avec un **objet**, on écrit `ninja.nom` et `ninja.chakra` : le nom dit ce que c'est. Et on peut donner au ninja des **actions** : `ninja.mediter(20)`, `ninja.lancer_jutsu()`.

> 💡 **L'idée de la POO :** regrouper dans une même « chose » ses **informations** (les attributs) et ce qu'elle **sait faire** (les méthodes).

# 2. Le vocabulaire

| Mot | Ce que c'est | Image | Exemple |
|:---|:---|:---|:---|
| **classe** | le **plan** qui décrit ce qu'est un ninja | le moule à gâteau | `Ninja` |
| **objet** (ou **instance**) | **un** ninja fabriqué avec ce plan | un gâteau | `kenji`, `aiko` |
| **attribut** | une **information** que possède chaque objet | la saveur du gâteau | `nom`, `village`, `chakra` |
| **méthode** | une **fonction écrite dans la classe** : ce que l'objet sait faire | | `mediter`, `lancer_jutsu` |
| **constructeur** | la méthode spéciale `__init__`, appelée **à la création** de l'objet : elle donne leur valeur de départ aux attributs | | `__init__` |

Avec **une** classe, on fabrique **autant d'objets qu'on veut**. Chaque objet a **ses propres** valeurs d'attributs.

> 📚 **Question type du bac** (Métropole 2023, sujet 0 2024) : « Citer un attribut et une méthode de la classe… » ou « Associer chaque nom au terme qui lui correspond parmi : objet, attribut, méthode ou classe ».
> - un **attribut** s'écrit `self.quelque_chose = ...` dans le constructeur ;
> - une **méthode** commence par `def` et est écrite **dans** la classe ;
> - la **classe** est le nom après `class` (avec une majuscule, par habitude) ;
> - un **objet** est créé par `nom = Classe(...)`.

# 3. Écrire une classe

Voici la classe `Ninja`. Les numéros de ligne servent pour les explications.
```python
1   class Ninja:
2       def __init__(self, nom, village, chakra):
3           self.nom = nom
4           self.village = village
5           self.chakra = chakra
6           self.nb_jutsus = 0
7
8       def mediter(self, quantite):
9           self.chakra = self.chakra + quantite
10
11      def lancer_jutsu(self):
12          self.chakra = self.chakra - 10
13          self.nb_jutsus = self.nb_jutsus + 1
```

- **Ligne 1 :** `class Ninja:` commence la classe. Tout ce qui est **décalé** en dessous fait partie de la classe.
- **Ligne 2 :** le **constructeur** `__init__` (deux tirets bas de chaque côté). Il reçoit `self`, puis les informations nécessaires pour fabriquer un ninja.
- **Lignes 3 à 5 :** on **range** les paramètres dans des attributs. `self.nom = nom` veut dire : « l'attribut `nom` de l'objet prend la valeur du paramètre `nom` ».
- **Ligne 6 :** un attribut qui ne vient **pas** d'un paramètre : tout ninja débutant n'a encore lancé aucun jutsu, il commence à 0. (Au bac : le solde d'un compte qui commence à 0, l'état d'un colis qui commence à `'préparé'`…)
- **Lignes 8 à 13 :** deux **méthodes**. Elles commencent **toujours** par le paramètre `self`.

### Et `self`, c'est quoi ?

`self` désigne **l'objet sur lequel on travaille**. Quand on écrit `kenji.lancer_jutsu()`, Python exécute `lancer_jutsu` avec `self = kenji` : c'est donc le chakra **de `kenji`** qui baisse, pas celui d'un autre ninja.

> 💡 **Dans la classe**, on accède à un attribut avec `self.attribut`. Sans `self.`, Python croit que c'est une simple variable de la fonction, qui disparaît à la fin de la méthode.

````python
class Ninja:
    def __init__(self, nom, village, chakra):
        self.nom = nom
        self.village = village
        self.chakra = chakra
        self.nb_jutsus = 0

    def mediter(self, quantite):
        self.chakra = self.chakra + quantite

    def lancer_jutsu(self):
        self.chakra = self.chakra - 10
        self.nb_jutsus = self.nb_jutsus + 1
````

# 4. Créer des objets et les utiliser

### Créer un objet

On écrit le **nom de la classe**, puis entre parenthèses les valeurs des paramètres du constructeur, **sans `self`** :
```python
kenji = Ninja("Kenji", "Vent", 50)
```
Python fabrique un nouveau ninja, appelle `__init__` avec `self` = ce nouveau ninja, `nom = "Kenji"`, `village = "Vent"` et `chakra = 50`, puis range l'objet dans la variable `kenji`.

> 📚 Au bac (Métropole 2022, Centres étrangers 2022), on demande de choisir la bonne écriture parmi plusieurs. La bonne est **toujours** de la forme `variable = Classe(valeurs)` : pas de `self` dans les parenthèses, pas de crochets, pas d'accolades.

### Lire un attribut, appeler une méthode

**En dehors de la classe**, on écrit le nom de l'objet, un point, puis l'attribut ou la méthode :
- `kenji.chakra` : la valeur de l'attribut `chakra` de `kenji` ;
- `kenji.lancer_jutsu()` : appelle la méthode `lancer_jutsu` sur `kenji` (les parenthèses sont obligatoires, même vides) ;
- `kenji.mediter(20)` : on donne la valeur du paramètre `quantite`, **sans** `self`.

````python
kenji = Ninja("Kenji", "Vent", 50)
aiko = Ninja("Aiko", "Rivière", 80)

kenji.lancer_jutsu()
kenji.lancer_jutsu()
aiko.mediter(15)

print(kenji.nom, kenji.chakra, kenji.nb_jutsus)
print(aiko.nom, aiko.chakra, aiko.nb_jutsus)
````

Les deux ninjas sont **indépendants** : quand `kenji` lance un jutsu, le chakra de `aiko` ne change pas.

| | dans la classe | en dehors de la classe |
|:---|:---|:---|
| lire un attribut | `self.chakra` | `kenji.chakra` |
| appeler une méthode | `self.lancer_jutsu()` | `kenji.lancer_jutsu()` |

# 5. Écrire une méthode : les 5 sortes

Dans **tous** les sujets du bac, les méthodes à écrire sont de l'une de ces 5 sortes. Avant d'écrire, on se demande : **laquelle est-ce ?**

### Sorte 1 — Renvoyer un attribut (un « accesseur », ou *getter*)
```python
def get_chakra(self):
    return self.chakra
```
📚 `get_note` (sujet 0 2024), `get_annee` (Asie 2025), `renvoie_nb_voisines` (Métropole 2023).

### Sorte 2 — Modifier un attribut (un « mutateur », ou *setter*)
```python
def set_village(self, nouveau_village):
    self.village = nouveau_village
```
Pas de `return` : la méthode **change** l'objet, elle ne renvoie rien.
📚 `crediter` / `debiter` (Nouvelle-Calédonie 2022), `changer_role` (Métropole 2024), `passer_transit` (Amérique du Nord 2025).

### Sorte 3 — Tester quelque chose (renvoie `True` ou `False`)
```python
def est_epuise(self):
    return self.chakra <= 0
```
Une comparaison vaut déjà `True` ou `False` : on peut la renvoyer directement, sans `if`.
📚 `est_positif` (Nouvelle-Calédonie 2022), `est_coloriee` (Métropole 2023), `est_bissextile` (Asie 2025).

### Sorte 4 — Utiliser une autre méthode de la classe
```python
def double_jutsu(self):
    self.lancer_jutsu()
    self.lancer_jutsu()
```
Dans la classe, on appelle une autre méthode avec `self.` devant.
📚 `transferer` qui utilise `debiter` et `crediter` (Nouvelle-Calédonie 2022), `newgame` qui utilise `raz` (Nouvelle-Calédonie 2022).

### Sorte 5 — Recevoir un **autre objet** en paramètre
```python
def transmettre_chakra(self, autre, quantite):
    self.chakra = self.chakra - quantite
    autre.mediter(quantite)
```
`self` est l'objet sur lequel on appelle la méthode, `autre` est l'objet passé en paramètre. On utilise `autre.attribut` ou `autre.methode()`.
📚 `transferer(self, autre_compte, montant)` (Nouvelle-Calédonie 2022), `est_superieure_a(self, autre)` (Polynésie 2025), `couvre(self, maison)` (Asie 2024), `attaquer(self, autre)` (Nouvelle-Calédonie 2022).

````python
class Ninja:
    def __init__(self, nom, village, chakra):
        self.nom = nom
        self.village = village
        self.chakra = chakra
        self.nb_jutsus = 0

    def mediter(self, quantite):
        self.chakra = self.chakra + quantite

    def lancer_jutsu(self):
        self.chakra = self.chakra - 10
        self.nb_jutsus = self.nb_jutsus + 1

    def get_chakra(self):                       # sorte 1
        return self.chakra

    def set_village(self, nouveau_village):     # sorte 2
        self.village = nouveau_village

    def est_epuise(self):                       # sorte 3
        return self.chakra <= 0

    def double_jutsu(self):                     # sorte 4
        self.lancer_jutsu()
        self.lancer_jutsu()

    def transmettre_chakra(self, autre, quantite):   # sorte 5
        self.chakra = self.chakra - quantite
        autre.mediter(quantite)


kenji = Ninja("Kenji", "Vent", 50)
aiko = Ninja("Aiko", "Rivière", 10)
kenji.transmettre_chakra(aiko, 30)
print(kenji.get_chakra(), aiko.get_chakra())
print(aiko.est_epuise())
````

# 6. Passer par les méthodes plutôt que toucher aux attributs

Certains sujets précisent : « **sans accéder directement aux attributs** » (Asie 2024, Nouvelle-Calédonie 2022). Cela veut dire : depuis l'extérieur d'une classe, on ne lit et on ne modifie pas `objet.attribut` ; on utilise les méthodes prévues.

| Interdit dans ce cas | Autorisé |
|:---|:---|
| `autre.chakra = autre.chakra + quantite` | `autre.mediter(quantite)` |
| `maison._position` | `maison.get_pos_maison()` |

**Pourquoi ?** Si un jour on change la façon dont la classe range ses informations, seul le code de la classe est à modifier : le reste du programme, qui passe par les méthodes, continue de marcher.

> 💡 Un attribut dont le nom commence par `_` (comme `self._position`) signale : « ne pas y toucher depuis l'extérieur, utiliser les méthodes ».

# 7. Une liste d'objets : les 5 modèles de parcours

Au bac, on range très souvent des objets dans une **liste** et on demande une fonction qui la parcourt. Il n'y a que **5 modèles**. On les montre sur une liste de ninjas.

````python
ninjas = [Ninja("Kenji", "Vent", 50), Ninja("Aiko", "Rivière", 0), Ninja("Ren", "Vent", 90)]
````

| Modèle | La question ressemble à… | Variable de départ | Dans la boucle | 📚 Au bac |
|:---|:---|:---|:---|:---|
| **1. compter** | « renvoie le **nombre** de… » | `nb = 0` | `nb = nb + 1` si la condition est vraie | `nb_maison` (Métropole 2021), `get_nb_rando` (Nouvelle-Calédonie 2025) |
| **2. additionner** | « renvoie le **total** de… » | `total = 0` | `total = total + objet.attribut` | `poids_total` (Amérique du Nord 2025), `temps_course` (Métropole 2024) |
| **3. filtrer** | « renvoie la **liste des**… qui… » | `res = []` | `res.append(objet)` si la condition est vraie | `liste_colis_etat` (Amérique du Nord 2025), `rechercher_debiteurs` (Nouvelle-Calédonie 2022) |
| **4. chercher** | « renvoie **le**… dont… », « renvoie `None` s'il n'existe pas » | — | `return` dès qu'on trouve ; `return None` après la boucle | `titre_livre` (sujet 0 2024) |
| **5. le meilleur** | « renvoie le **plus**… » | le premier objet | on garde l'objet s'il est meilleur | `meilleur_concurrent` (Amérique du Nord 2022), `renvoie_max` (Métropole 2023) |

````python
def nb_du_village(liste, village):     # 1. compter
    nb = 0
    for n in liste:
        if n.village == village:
            nb = nb + 1
    return nb


def chakra_total(liste):               # 2. additionner
    total = 0
    for n in liste:
        total = total + n.chakra
    return total


def ninjas_epuises(liste):             # 3. filtrer
    res = []
    for n in liste:
        if n.est_epuise():
            res.append(n)
    return res


def chercher_nom(liste, nom):          # 4. chercher
    for n in liste:
        if n.nom == nom:
            return n
    return None


def plus_fort(liste):                  # 5. le meilleur
    meilleur = liste[0]
    for n in liste:
        if n.chakra > meilleur.chakra:
            meilleur = n
    return meilleur


print(nb_du_village(ninjas, "Vent"))
print(chakra_total(ninjas))
print(len(ninjas_epuises(ninjas)))
print(chercher_nom(ninjas, "Aiko").chakra)
print(chercher_nom(ninjas, "Hana"))
print(plus_fort(ninjas).nom)
````

Le modèle 3 s'écrit aussi en **une ligne**, avec une liste en compréhension. Les deux écritures font exactement la même chose :

| Avec une boucle | Avec une liste en compréhension |
|:---|:---|
| `res = []`<br>`for n in liste:`<br>`    if n.est_epuise():`<br>`        res.append(n)` | `res = [n for n in liste if n.est_epuise()]` |

> ⚠️ Dans la boucle `for n in liste`, la variable `n` est **un objet** : on écrit `n.chakra`, pas `liste[n]` ni `n[2]`.

# 8. Un objet qui contient d'autres objets

Un attribut peut contenir **une liste d'objets**. C'est le cas dans presque tous les gros exercices : une `Bibliotheque` contient des `Livre` (sujet 0 2024), une `Equipe` contient des `Chien` (Métropole 2024), un `Paquet` contient des `Carte` (Polynésie 2025).

Ici, une **équipe** de ninjas contient plusieurs objets `Ninja` :

````python
class Equipe:
    def __init__(self, nom):
        self.nom = nom
        self.membres = []               # une liste vide de ninjas

    def ajouter(self, ninja):
        self.membres.append(ninja)

    def nb_membres(self):
        return len(self.membres)

    def entrainer_tous(self, quantite):
        for n in self.membres:          # n est un objet Ninja
            n.mediter(quantite)         # on utilise une méthode de Ninja


equipe = Equipe("Équipe 3")
equipe.ajouter(Ninja("Kenji", "Vent", 50))
equipe.ajouter(Ninja("Aiko", "Rivière", 0))
equipe.entrainer_tous(10)
print(equipe.nb_membres())
for n in equipe.membres:
    print(n.nom, n.chakra)
````

Dans une méthode d'`Equipe`, on parcourt `self.membres` avec **les mêmes 5 modèles** que dans la partie 7 : la seule différence est qu'on écrit `self.membres` au lieu de `liste`.

# 9. Suivre un programme avec des objets (tableau de trace)

Pour savoir ce qu'affiche un programme qui utilise des objets, on fait un tableau de trace avec **une colonne par attribut de chaque objet**, et **une ligne par instruction**.

Exemple :
```python
kenji = Ninja("Kenji", "Vent", 50)
aiko = Ninja("Aiko", "Rivière", 10)
kenji.lancer_jutsu()
kenji.transmettre_chakra(aiko, 30)
aiko.lancer_jutsu()
print(kenji.chakra, aiko.chakra, aiko.nb_jutsus)
```

| Instruction | `kenji.chakra` | `kenji.nb_jutsus` | `aiko.chakra` | `aiko.nb_jutsus` |
|:---|:---:|:---:|:---:|:---:|
| `kenji = Ninja("Kenji", "Vent", 50)` | 50 | 0 | — | — |
| `aiko = Ninja("Aiko", "Rivière", 10)` | 50 | 0 | 10 | 0 |
| `kenji.lancer_jutsu()` | 40 | 1 | 10 | 0 |
| `kenji.transmettre_chakra(aiko, 30)` | 10 | 1 | 40 | 0 |
| `aiko.lancer_jutsu()` | 10 | 1 | 30 | 1 |

Le programme affiche `10 30 1`.

> 💡 Quand une méthode en appelle une autre (ici `transmettre_chakra` appelle `mediter`), on **ouvre** la méthode appelée et on la suit jusqu'au bout avant de passer à l'instruction suivante. C'est là que se cachent les surprises (voir l'exercice des chevaliers).

````python
# Vérification
kenji = Ninja("Kenji", "Vent", 50)
aiko = Ninja("Aiko", "Rivière", 10)
kenji.lancer_jutsu()
kenji.transmettre_chakra(aiko, 30)
aiko.lancer_jutsu()
print(kenji.chakra, aiko.chakra, aiko.nb_jutsus)
````

# 10. La méthode pour résoudre un exercice de POO

Face à une question, on suit toujours les mêmes étapes, **sur papier** :

1. **Repérer les classes** de l'énoncé et, pour chacune, **lister ses attributs** (dans le constructeur) et **ses méthodes**. Les noms exacts servent à chaque question.
2. **Dire où on écrit** : **dans** la classe (une méthode : `self`) ou **en dehors** (une fonction ou des instructions : `objet.`) ?
3. **Reconnaître la sorte** :
   - une méthode → une des 5 sortes de la partie 5 ;
   - une fonction sur une liste d'objets → un des 5 modèles de la partie 7.
4. **Écrire le squelette** de la sorte ou du modèle, puis le **remplir** avec les noms de l'énoncé.
5. **Vérifier sur un petit exemple** de l'énoncé, avec un tableau de trace si besoin.

> 💡 Au bac, l'énoncé dit souvent : « on pourra utiliser les méthodes des questions précédentes, même si elles n'ont pas été écrites ». Il faut le faire : c'est souvent la clé de la question.

# 11. Fiche réflexe

| On te demande… | Tu écris… |
|:---|:---|
| de **créer** un objet | `nom = Classe(valeur1, valeur2)` (sans `self`) |
| de **compléter le constructeur** | `self.attribut = parametre` pour chaque paramètre, et les valeurs de départ fixes (`self.solde = 0`) |
| la **valeur d'un attribut** | `objet.attribut` (en dehors) ou `self.attribut` (dans la classe) |
| d'**appeler** une méthode | `objet.methode(valeurs)` (sans `self`) |
| une méthode qui **renvoie** un attribut | `return self.attribut` |
| une méthode qui **modifie** | `self.attribut = ...`, sans `return` |
| une méthode qui **teste** | `return condition` |
| une méthode qui reçoit un **autre objet** | `autre.attribut` ou `autre.methode()` |
| une fonction sur une **liste d'objets** | un des 5 modèles : compter, additionner, filtrer, chercher, le meilleur |
| ce qu'**affiche** un programme | tableau de trace : une colonne par attribut de chaque objet |
| « citer un attribut / une méthode » | attribut : `self.xxx = ...` dans `__init__` ; méthode : un `def` dans la classe |
