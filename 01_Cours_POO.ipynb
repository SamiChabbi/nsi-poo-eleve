# 1. Pourquoi des objets ?

Un téléphone a une **marque** et un niveau de **batterie**. Sans objets, on pourrait le ranger dans une liste :

````python
tel = ["Nokia", 50]
print(tel[0])   # la marque
print(tel[1])   # la batterie
````

Ça marche, mais `tel[1]` ne dit pas ce que c'est : il faut se souvenir que la case 1 est la batterie. Et rien n'empêche d'écrire n'importe quoi dedans.

Avec un **objet**, on écrit `tel.marque` et `tel.batterie` : le nom dit ce que c'est. Et on peut donner au téléphone des **actions** : `tel.recharger(20)`, `tel.appeler()`.

> 💡 **L'idée de la POO :** regrouper dans une même « chose » ses **informations** (les attributs) et ce qu'elle **sait faire** (les méthodes).

# 2. Le vocabulaire

| Mot | Ce que c'est | Image | Exemple |
|:---|:---|:---|:---|
| **classe** | le **plan** qui décrit ce qu'est un téléphone | le moule à gâteau | `Telephone` |
| **objet** (ou **instance**) | **un** téléphone fabriqué avec ce plan | un gâteau | `tel1`, `tel2` |
| **attribut** | une **information** que possède chaque objet | la saveur du gâteau | `marque`, `batterie` |
| **méthode** | une **fonction écrite dans la classe** : ce que l'objet sait faire | | `recharger`, `appeler` |
| **constructeur** | la méthode spéciale `__init__`, appelée **à la création** de l'objet : elle donne leur valeur de départ aux attributs | | `__init__` |

Avec **une** classe, on fabrique **autant d'objets qu'on veut**. Chaque objet a **ses propres** valeurs d'attributs.

> 📚 **Question type du bac** (Métropole 2023, sujet 0 2024) : « Citer un attribut et une méthode de la classe… » ou « Associer chaque nom au terme qui lui correspond parmi : objet, attribut, méthode ou classe ».
> - un **attribut** s'écrit `self.quelque_chose = ...` dans le constructeur ;
> - une **méthode** commence par `def` et est écrite **dans** la classe ;
> - la **classe** est le nom après `class` (avec une majuscule, par habitude) ;
> - un **objet** est créé par `nom = Classe(...)`.

# 3. Écrire une classe

Voici la classe `Telephone`. Les numéros de ligne servent pour les explications.
```python
1   class Telephone:
2       def __init__(self, marque, batterie):
3           self.marque = marque
4           self.batterie = batterie
5           self.nb_appels = 0
6
7       def recharger(self, quantite):
8           self.batterie = self.batterie + quantite
9
10      def appeler(self):
11          self.batterie = self.batterie - 10
12          self.nb_appels = self.nb_appels + 1
```

- **Ligne 1 :** `class Telephone:` commence la classe. Tout ce qui est **décalé** en dessous fait partie de la classe.
- **Ligne 2 :** le **constructeur** `__init__` (deux tirets bas de chaque côté). Il reçoit `self`, puis les informations nécessaires pour fabriquer un téléphone.
- **Lignes 3 et 4 :** on **range** les paramètres dans des attributs. `self.marque = marque` veut dire : « l'attribut `marque` de l'objet prend la valeur du paramètre `marque` ».
- **Ligne 5 :** un attribut qui ne vient **pas** d'un paramètre : tout téléphone neuf commence avec 0 appel. (Au bac : le solde d'un compte qui commence à 0, l'état d'un colis qui commence à `'préparé'`…)
- **Lignes 7 à 12 :** deux **méthodes**. Elles commencent **toujours** par le paramètre `self`.

### Et `self`, c'est quoi ?

`self` désigne **l'objet sur lequel on travaille**. Quand on écrit `tel1.appeler()`, Python exécute `appeler` avec `self = tel1` : c'est donc la batterie **de `tel1`** qui baisse, pas celle d'un autre téléphone.

> 💡 **Dans la classe**, on accède à un attribut avec `self.attribut`. Sans `self.`, Python croit que c'est une simple variable de la fonction, qui disparaît à la fin de la méthode.

````python
class Telephone:
    def __init__(self, marque, batterie):
        self.marque = marque
        self.batterie = batterie
        self.nb_appels = 0

    def recharger(self, quantite):
        self.batterie = self.batterie + quantite

    def appeler(self):
        self.batterie = self.batterie - 10
        self.nb_appels = self.nb_appels + 1
````

# 4. Créer des objets et les utiliser

### Créer un objet

On écrit le **nom de la classe**, puis entre parenthèses les valeurs des paramètres du constructeur, **sans `self`** :
```python
tel1 = Telephone("Nokia", 50)
```
Python fabrique un nouveau téléphone, appelle `__init__` avec `self` = ce nouveau téléphone, `marque = "Nokia"` et `batterie = 50`, puis range l'objet dans la variable `tel1`.

> 📚 Au bac (Métropole 2022, Centres étrangers 2022), on demande de choisir la bonne écriture parmi plusieurs. La bonne est **toujours** de la forme `variable = Classe(valeurs)` : pas de `self` dans les parenthèses, pas de crochets, pas d'accolades.

### Lire un attribut, appeler une méthode

**En dehors de la classe**, on écrit le nom de l'objet, un point, puis l'attribut ou la méthode :
- `tel1.batterie` : la valeur de l'attribut `batterie` de `tel1` ;
- `tel1.appeler()` : appelle la méthode `appeler` sur `tel1` (les parenthèses sont obligatoires, même vides) ;
- `tel1.recharger(20)` : on donne la valeur du paramètre `quantite`, **sans** `self`.

````python
tel1 = Telephone("Nokia", 50)
tel2 = Telephone("Fairphone", 80)

tel1.appeler()
tel1.appeler()
tel2.recharger(15)

print(tel1.marque, tel1.batterie, tel1.nb_appels)
print(tel2.marque, tel2.batterie, tel2.nb_appels)
````

Les deux téléphones sont **indépendants** : appeler avec `tel1` ne change rien à `tel2`.

| | dans la classe | en dehors de la classe |
|:---|:---|:---|
| lire un attribut | `self.batterie` | `tel1.batterie` |
| appeler une méthode | `self.appeler()` | `tel1.appeler()` |

# 5. Écrire une méthode : les 5 sortes

Dans **tous** les sujets du bac, les méthodes à écrire sont de l'une de ces 5 sortes. Avant d'écrire, on se demande : **laquelle est-ce ?**

### Sorte 1 — Renvoyer un attribut (un « accesseur », ou *getter*)
```python
def get_batterie(self):
    return self.batterie
```
📚 `get_note` (sujet 0 2024), `get_annee` (Asie 2025), `renvoie_nb_voisines` (Métropole 2023).

### Sorte 2 — Modifier un attribut (un « mutateur », ou *setter*)
```python
def set_marque(self, nouvelle_marque):
    self.marque = nouvelle_marque
```
Pas de `return` : la méthode **change** l'objet, elle ne renvoie rien.
📚 `crediter` / `debiter` (Nouvelle-Calédonie 2022), `changer_role` (Métropole 2024), `passer_transit` (Amérique du Nord 2025).

### Sorte 3 — Tester quelque chose (renvoie `True` ou `False`)
```python
def est_vide(self):
    return self.batterie <= 0
```
Une comparaison vaut déjà `True` ou `False` : on peut la renvoyer directement, sans `if`.
📚 `est_positif` (Nouvelle-Calédonie 2022), `est_coloriee` (Métropole 2023), `est_bissextile` (Asie 2025).

### Sorte 4 — Utiliser une autre méthode de la classe
```python
def appeler_deux_fois(self):
    self.appeler()
    self.appeler()
```
Dans la classe, on appelle une autre méthode avec `self.` devant.
📚 `transferer` qui utilise `debiter` et `crediter` (Nouvelle-Calédonie 2022), `newgame` qui utilise `raz` (Nouvelle-Calédonie 2022).

### Sorte 5 — Recevoir un **autre objet** en paramètre
```python
def donner_batterie(self, autre, quantite):
    self.batterie = self.batterie - quantite
    autre.recharger(quantite)
```
`self` est l'objet sur lequel on appelle la méthode, `autre` est l'objet passé en paramètre. On utilise `autre.attribut` ou `autre.methode()`.
📚 `transferer(self, autre_compte, montant)` (Nouvelle-Calédonie 2022), `est_superieure_a(self, autre)` (Polynésie 2025), `couvre(self, maison)` (Asie 2024), `attaquer(self, autre)` (Nouvelle-Calédonie 2022).

````python
class Telephone:
    def __init__(self, marque, batterie):
        self.marque = marque
        self.batterie = batterie
        self.nb_appels = 0

    def recharger(self, quantite):
        self.batterie = self.batterie + quantite

    def appeler(self):
        self.batterie = self.batterie - 10
        self.nb_appels = self.nb_appels + 1

    def get_batterie(self):                    # sorte 1
        return self.batterie

    def set_marque(self, nouvelle_marque):     # sorte 2
        self.marque = nouvelle_marque

    def est_vide(self):                        # sorte 3
        return self.batterie <= 0

    def appeler_deux_fois(self):               # sorte 4
        self.appeler()
        self.appeler()

    def donner_batterie(self, autre, quantite):   # sorte 5
        self.batterie = self.batterie - quantite
        autre.recharger(quantite)


tel1 = Telephone("Nokia", 50)
tel2 = Telephone("Fairphone", 10)
tel1.donner_batterie(tel2, 30)
print(tel1.get_batterie(), tel2.get_batterie())
print(tel2.est_vide())
````

# 6. Passer par les méthodes plutôt que toucher aux attributs

Certains sujets précisent : « **sans accéder directement aux attributs** » (Asie 2024, Nouvelle-Calédonie 2022). Cela veut dire : depuis l'extérieur d'une classe, on ne lit et on ne modifie pas `objet.attribut` ; on utilise les méthodes prévues.

| Interdit dans ce cas | Autorisé |
|:---|:---|
| `autre.batterie = autre.batterie + quantite` | `autre.recharger(quantite)` |
| `maison._position` | `maison.get_pos_maison()` |

**Pourquoi ?** Si un jour on change la façon dont la classe range ses informations, seul le code de la classe est à modifier : le reste du programme, qui passe par les méthodes, continue de marcher.

> 💡 Un attribut dont le nom commence par `_` (comme `self._position`) signale : « ne pas y toucher depuis l'extérieur, utiliser les méthodes ».

# 7. Une liste d'objets : les 5 modèles de parcours

Au bac, on range très souvent des objets dans une **liste** et on demande une fonction qui la parcourt. Il n'y a que **5 modèles**. On les montre sur une liste de téléphones.

````python
telephones = [Telephone("Nokia", 50), Telephone("Fairphone", 0), Telephone("Nokia", 90)]
````

| Modèle | La question ressemble à… | Variable de départ | Dans la boucle | 📚 Au bac |
|:---|:---|:---|:---|:---|
| **1. compter** | « renvoie le **nombre** de… » | `nb = 0` | `nb = nb + 1` si la condition est vraie | `nb_maison` (Métropole 2021), `get_nb_rando` (Nouvelle-Calédonie 2025) |
| **2. additionner** | « renvoie le **total** de… » | `total = 0` | `total = total + objet.attribut` | `poids_total` (Amérique du Nord 2025), `temps_course` (Métropole 2024) |
| **3. filtrer** | « renvoie la **liste des**… qui… » | `res = []` | `res.append(objet)` si la condition est vraie | `liste_colis_etat` (Amérique du Nord 2025), `rechercher_debiteurs` (Nouvelle-Calédonie 2022) |
| **4. chercher** | « renvoie **le**… dont… », « renvoie `None` s'il n'existe pas » | — | `return` dès qu'on trouve ; `return None` après la boucle | `titre_livre` (sujet 0 2024) |
| **5. le meilleur** | « renvoie le **plus**… » | le premier objet | on garde l'objet s'il est meilleur | `meilleur_concurrent` (Amérique du Nord 2022), `renvoie_max` (Métropole 2023) |

````python
def nb_nokia(liste):                  # 1. compter
    nb = 0
    for t in liste:
        if t.marque == "Nokia":
            nb = nb + 1
    return nb


def batterie_totale(liste):           # 2. additionner
    total = 0
    for t in liste:
        total = total + t.batterie
    return total


def telephones_vides(liste):          # 3. filtrer
    res = []
    for t in liste:
        if t.est_vide():
            res.append(t)
    return res


def chercher_marque(liste, marque):   # 4. chercher
    for t in liste:
        if t.marque == marque:
            return t
    return None


def plus_charge(liste):               # 5. le meilleur
    meilleur = liste[0]
    for t in liste:
        if t.batterie > meilleur.batterie:
            meilleur = t
    return meilleur


print(nb_nokia(telephones))
print(batterie_totale(telephones))
print(len(telephones_vides(telephones)))
print(chercher_marque(telephones, "Fairphone").batterie)
print(chercher_marque(telephones, "Apple"))
print(plus_charge(telephones).batterie)
````

Le modèle 3 s'écrit aussi en **une ligne**, avec une liste en compréhension. Les deux écritures font exactement la même chose :

| Avec une boucle | Avec une liste en compréhension |
|:---|:---|
| `res = []`<br>`for t in liste:`<br>`    if t.est_vide():`<br>`        res.append(t)` | `res = [t for t in liste if t.est_vide()]` |

> ⚠️ Dans la boucle `for t in liste`, la variable `t` est **un objet** : on écrit `t.batterie`, pas `liste[t]` ni `t[1]`.

# 8. Un objet qui contient d'autres objets

Un attribut peut contenir **une liste d'objets**. C'est le cas dans presque tous les gros exercices : une `Bibliotheque` contient des `Livre` (sujet 0 2024), une `Equipe` contient des `Chien` (Métropole 2024), un `Paquet` contient des `Carte` (Polynésie 2025).

````python
class Magasin:
    def __init__(self, nom):
        self.nom = nom
        self.stock = []                 # une liste vide de téléphones

    def ajouter(self, telephone):
        self.stock.append(telephone)

    def nb_telephones(self):
        return len(self.stock)

    def tout_recharger(self, quantite):
        for t in self.stock:            # t est un objet Telephone
            t.recharger(quantite)       # on utilise une méthode de Telephone


m = Magasin("Le coin du mobile")
m.ajouter(Telephone("Nokia", 50))
m.ajouter(Telephone("Fairphone", 0))
m.tout_recharger(10)
print(m.nb_telephones())
for t in m.stock:
    print(t.marque, t.batterie)
````

Dans une méthode de `Magasin`, on parcourt `self.stock` avec **les mêmes 5 modèles** que dans la partie 7 : la seule différence est qu'on écrit `self.stock` au lieu de `liste`.

# 9. Suivre un programme avec des objets (tableau de trace)

Pour savoir ce qu'affiche un programme qui utilise des objets, on fait un tableau de trace avec **une colonne par attribut de chaque objet**, et **une ligne par instruction**.

Exemple :
```python
tel1 = Telephone("Nokia", 50)
tel2 = Telephone("Fairphone", 10)
tel1.appeler()
tel1.donner_batterie(tel2, 30)
tel2.appeler()
print(tel1.batterie, tel2.batterie, tel2.nb_appels)
```

| Instruction | `tel1.batterie` | `tel1.nb_appels` | `tel2.batterie` | `tel2.nb_appels` |
|:---|:---:|:---:|:---:|:---:|
| `tel1 = Telephone("Nokia", 50)` | 50 | 0 | — | — |
| `tel2 = Telephone("Fairphone", 10)` | 50 | 0 | 10 | 0 |
| `tel1.appeler()` | 40 | 1 | 10 | 0 |
| `tel1.donner_batterie(tel2, 30)` | 10 | 1 | 40 | 0 |
| `tel2.appeler()` | 10 | 1 | 30 | 1 |

Le programme affiche `10 30 1`.

> 💡 Quand une méthode en appelle une autre (ici `donner_batterie` appelle `recharger`), on **ouvre** la méthode appelée et on la suit jusqu'au bout avant de passer à l'instruction suivante. C'est là que se cachent les surprises (voir l'exercice des chevaliers).

````python
# Vérification
tel1 = Telephone("Nokia", 50)
tel2 = Telephone("Fairphone", 10)
tel1.appeler()
tel1.donner_batterie(tel2, 30)
tel2.appeler()
print(tel1.batterie, tel2.batterie, tel2.nb_appels)
````

# 10. La méthode pour résoudre un exercice de POO

   - une méthode → une des 5 sortes de la partie 5 ;
   - une fonction sur une liste d'objets → un des 5 modèles de la partie 7.
4. **Écrire le squelette** de la sorte ou du modèle, puis le **remplir** avec les noms de l'énoncé.

> 💡 Au bac, l'énoncé dit souvent : « on pourra utiliser les méthodes des questions précédentes, même si elles n'ont pas été écrites ». Il faut le faire : c'est souvent la clé de la question.

# 11. Fiche réflexe

| On te demande… | Tu écris… |
|:---|:---|
| de **créer** un objet | `nom = Classe(valeur1, valeur2)` (sans `self`) |
| de **compléter le constructeur** | `self.attribut = parametre` pour chaque paramètre, et les valeurs de départ fixes (`self.solde = 0`) |

| une méthode qui **renvoie** un attribut | `return self.attribut` |
| une méthode qui **modifie** | `self.attribut = ...`, sans `return` |
| une méthode qui **teste** | `return condition` |
| une méthode qui reçoit un **autre objet** | `autre.attribut` ou `autre.methode()` |
| une fonction sur une **liste d'objets** | un des 5 modèles : compter, additionner, filtrer, chercher, le meilleur |
| ce qu'**affiche** un programme | tableau de trace : une colonne par attribut de chaque objet |
| « citer un attribut / une méthode » | attribut : `self.xxx = ...` dans `__init__` ; méthode : un `def` dans la classe |
